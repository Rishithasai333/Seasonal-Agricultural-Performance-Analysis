```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

import warnings
warnings.filterwarnings("ignore")

print("Libraries imported successfully!")
```

```python
uploaded = files.upload()

file_name = list(uploaded.keys())[0]

if file_name.endswith(".xlsx"):
    df = pd.read_excel(file_name)
elif file_name.endswith(".csv"):
    df = pd.read_csv(file_name)
else:
    raise ValueError("Please upload an Excel or CSV file")

print("Dataset loaded successfully!")
print("File:", file_name)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
for i, col in enumerate(df.columns, 1):
    print(i, "-", col)

print("\nData Types:")
display(df.dtypes)

print("\nStatistical Summary:")
display(df.describe())

print("\nMissing Values:")
display(df.isnull().sum())
```

```python
# CELL 4: Data Cleaning

print("Duplicate rows before cleaning:", df.duplicated().sum())

df = df.drop_duplicates()

numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

categorical_columns = df.select_dtypes(include="object").columns

for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Duplicate rows after cleaning:", df.duplicated().sum())

print("\nMissing values after cleaning:")
display(df.isnull().sum())

print("\nCleaned Dataset Shape:", df.shape)
```

```python
print("Seasons in Dataset:")
print(df["Season"].value_counts())

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Season")

plt.title("Distribution of Agricultural Activities by Season")
plt.xlabel("Season")
plt.ylabel("Number of Farms")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
seasonal_performance = df.groupby("Season")[
    [
        "Yield_Tonnes_Ha",
        "Production_Tonnes",
        "Revenue_INR",
        "Profit_INR",
        "Water_Efficiency_t_per_1000m3"
    ]
].mean().round(2)

print("Average Agricultural Performance by Season:")

display(seasonal_performance)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(
    data=df,
    x="Season",
    y="Yield_Tonnes_Ha",
    estimator="mean",
    ax=axes[0]
)

axes[0].set_title("Average Yield Across Seasons")
axes[0].set_xlabel("Season")
axes[0].set_ylabel("Yield (Tonnes/Ha)")
axes[0].tick_params(axis="x", rotation=45)

sns.barplot(
    data=df,
    x="Season",
    y="Profit_INR",
    estimator="mean",
    ax=axes[1]
)

axes[1].set_title("Average Profit Across Seasons")
axes[1].set_xlabel("Season")
axes[1].set_ylabel("Profit (INR)")
axes[1].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```

```python
environment_columns = [
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Sunlight_Hours_Day",
    "Soil_Moisture_pct",
    "Yield_Tonnes_Ha"
]

correlation = df[environment_columns].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Environmental Conditions vs Agricultural Yield")

plt.tight_layout()
plt.show()
```

```python
resource_summary = df.groupby("Season")[
    [
        "Water_Used_m3",
        "Water_Efficiency_t_per_1000m3",
        "Fertilizer_kg_ha",
        "Pesticide_Litre_ha",
        "Nitrogen_kg_ha",
        "Phosphorus_kg_ha",
        "Potassium_kg_ha"
    ]
].mean().round(2)

print("Average Resource Usage by Season:")

display(resource_summary)

print("\nIrrigation Method Distribution:")

irrigation = pd.crosstab(
    df["Season"],
    df["Irrigation_Method"]
)

display(irrigation)

irrigation.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Irrigation Methods Across Seasons")
plt.xlabel("Season")
plt.ylabel("Number of Farms")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
print("=" * 70)
print("SEASONAL AGRICULTURE PERFORMANCE ANALYSIS")
print("=" * 70)

yield_by_season = df.groupby("Season")["Yield_Tonnes_Ha"].mean()
profit_by_season = df.groupby("Season")["Profit_INR"].mean()
water_efficiency = df.groupby("Season")[
    "Water_Efficiency_t_per_1000m3"
].mean()

best_yield_season = yield_by_season.idxmax()
worst_yield_season = yield_by_season.idxmin()

best_profit_season = profit_by_season.idxmax()
worst_profit_season = profit_by_season.idxmin()

best_water_season = water_efficiency.idxmax()

print("\nKEY FINDINGS")
print("-" * 70)

print(
    f"1. The highest average yield was observed in "
    f"{best_yield_season}."
)

print(
    f"2. The lowest average yield was observed in "
    f"{worst_yield_season}."
)

print(
    f"3. The highest average profit was observed in "
    f"{best_profit_season}."
)

print(
    f"4. The lowest average profit was observed in "
    f"{worst_profit_season}."
)

print(
    f"5. The highest water-use efficiency was observed in "
    f"{best_water_season}."
)

risk_by_season = df.groupby("Season")[
    "Disease_Pest_Risk_pct"
].mean()

highest_risk_season = risk_by_season.idxmax()

print(
    f"6. The highest average disease/pest risk was observed in "
    f"{highest_risk_season}."
)

print("\nCONCLUSION")
print("-" * 70)

print(
    "Agricultural performance varies across seasons. "
    "Seasonal conditions influence yield, production, profit, "
    "resource usage and water efficiency. The analysis helps "
    "identify better-performing seasons and areas requiring "
    "further investigation."
)

print("\nRECOMMENDATIONS")
print("-" * 70)

print(
    "1. Plan crop activities according to high-performing seasons."
)

print(
    "2. Optimize irrigation and water usage based on seasonal conditions."
)

print(
    "3. Monitor rainfall, temperature, humidity and soil moisture."
)

print(
    "4. Optimize fertilizer and pesticide usage to improve productivity."
)

print(
    "5. Give special attention to seasons with higher disease/pest risk."
)

print(
    "6. Use seasonal data analysis to support evidence-based "
    "agricultural planning."
)

print("\nAnalysis Completed Successfully!")
```
